```python
# E-Commerce Customer Behavior Analysis
# Dataset: UCI Online Retail — 541k transactions, Dec 2010 to Dec 2011
# UK-based wholesale gift retailer

# ── 1. Load Data ──────────────────────────────────────────────────────────

import zipfile

zip_path     = '/content/drive/MyDrive/data/online+retail.zip'
extract_path = '/content/drive/MyDrive/data/'

with zipfile.ZipFile(zip_path, 'r') as z:
    z.extractall(extract_path)
    print(z.namelist())

import pandas as pd

df = pd.read_excel('/content/drive/MyDrive/data/Online Retail.xlsx')

print("Shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nFirst 5 rows:")
df.head()

# ── 2. Data Quality Check ─────────────────────────────────────────────────
# Before cleaning, I check data types, missing values, and obvious anomalies

print("=== Data Types ===")
print(df.dtypes)

print("\n=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Basic Stats ===")
print(df[['Quantity', 'UnitPrice']].describe())

print("\n=== Cancelled Orders ===")
cancelled = df[df['InvoiceNo'].astype(str).str.startswith('C')]
print(f"Cancelled transactions: {len(cancelled)}")

print("\n=== Negative Quantities ===")
print(f"Rows with negative quantity: {(df['Quantity'] < 0).sum()}")

print("\n=== Unique counts ===")
print(f"Customers: {df['CustomerID'].nunique()}")
print(f"Countries: {df['Country'].nunique()}")
print(f"Products:  {df['StockCode'].nunique()}")
print(f"Invoices:  {df['InvoiceNo'].nunique()}")

# ── 3. Data Cleaning ──────────────────────────────────────────────────────

print(f"Rows before cleaning: {len(df)}")

# Rows without a CustomerID can't be used for customer-level analysis
df = df.dropna(subset=['CustomerID'])
print(f"After dropping missing CustomerID: {len(df)}")

# Drop the small number of rows with no product description
df = df.dropna(subset=['Description'])
print(f"After dropping missing Description: {len(df)}")

# Invoices starting with 'C' are cancellations/refunds, not purchases
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
print(f"After removing cancellations: {len(df)}")

# Remove returns and data errors — valid sales must have positive quantity
df = df[df['Quantity'] > 0]
print(f"After removing negative/zero quantity: {len(df)}")

# Remove rows with zero or negative price — these are data entry errors
df = df[df['UnitPrice'] > 0]
print(f"After removing negative/zero price: {len(df)}")

# CustomerID comes in as float (e.g. 17850.0) — convert to string identifier
df['CustomerID'] = df['CustomerID'].astype(int).astype(str)

# Revenue per line item
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']

print(f"\nFinal clean dataset: {len(df)} rows")
print(f"Removed {541909 - len(df)} rows total ({(541909 - len(df))/541909*100:.1f}%)")
df.head()

# ── 4. Exploratory Analysis ───────────────────────────────────────────────

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid')

# Revenue by country — grouped and summed, keeping only top 10
revenue_by_country = (df.groupby('Country')['TotalPrice']
                        .sum()
                        .nlargest(10)
                        .reset_index())

plt.figure(figsize=(12, 5))
sns.barplot(data=revenue_by_country, x='Country', y='TotalPrice',
            palette='Blues_d', legend=False)
plt.title('Top 10 Countries by Revenue', fontsize=14)
plt.xlabel('Country')
plt.ylabel('Total Revenue (£)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print(revenue_by_country)

# Monthly revenue — I extract year-month from the timestamp and group by it
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M')

monthly_revenue = (df.groupby('YearMonth')['TotalPrice']
                     .sum()
                     .reset_index())
monthly_revenue['YearMonth'] = monthly_revenue['YearMonth'].astype(str)

plt.figure(figsize=(14, 5))
plt.plot(monthly_revenue['YearMonth'], monthly_revenue['TotalPrice'],
         marker='o', linewidth=2, color='steelblue', markersize=5)
plt.fill_between(range(len(monthly_revenue)), monthly_revenue['TotalPrice'],
                 alpha=0.3, color='steelblue')
plt.title('Monthly Revenue Over Time', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Total Revenue (£)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print(monthly_revenue)

# Orders by day and hour — I count unique invoices (not rows) per period
# since one order can contain many product lines
df['DayOfWeek'] = df['InvoiceDate'].dt.dayofweek
df['Hour']      = df['InvoiceDate'].dt.hour

day_names = {0:'Monday', 1:'Tuesday', 2:'Wednesday',
             3:'Thursday', 4:'Friday', 5:'Saturday', 6:'Sunday'}
df['DayName'] = df['DayOfWeek'].map(day_names)

orders_by_day = (df.groupby('DayName')['InvoiceNo']
                   .nunique()
                   .reindex(['Monday','Tuesday','Wednesday',
                             'Thursday','Friday','Saturday','Sunday'])
                   .reset_index())
orders_by_day.columns = ['DayName', 'OrderCount']

orders_by_hour = (df.groupby('Hour')['InvoiceNo']
                    .nunique()
                    .reset_index())
orders_by_hour.columns = ['Hour', 'OrderCount']

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sns.barplot(data=orders_by_day,  x='DayName', y='OrderCount',
            palette='Blues_d', ax=axes[0])
axes[0].set_title('Orders by Day of Week', fontsize=13)
axes[0].set_xlabel('Day')
axes[0].set_ylabel('Number of Orders')
axes[0].tick_params(axis='x', rotation=45)

sns.barplot(data=orders_by_hour, x='Hour', y='OrderCount',
            palette='Blues_d', ax=axes[1])
axes[1].set_title('Orders by Hour of Day', fontsize=13)
axes[1].set_xlabel('Hour')
axes[1].set_ylabel('Number of Orders')

plt.tight_layout()
plt.show()

print("=== Orders by Day ===")
print(orders_by_day)
print("\n=== Orders by Hour ===")
print(orders_by_hour)

# Top products by revenue and quantity — comparing the two lists shows which
# products are genuinely valuable vs just high volume at low price
top_revenue = (df.groupby('Description')['TotalPrice']
                 .sum()
                 .nlargest(10)
                 .reset_index())
top_revenue.columns = ['Product', 'Revenue']

top_quantity = (df.groupby('Description')['Quantity']
                  .sum()
                  .nlargest(10)
                  .reset_index())
top_quantity.columns = ['Product', 'QuantitySold']

fig, axes = plt.subplots(1, 2, figsize=(18, 5))

sns.barplot(data=top_revenue,  x='Revenue',      y='Product',
            palette='Blues_d', ax=axes[0])
axes[0].set_title('Top 10 Products by Revenue', fontsize=13)
axes[0].set_xlabel('Total Revenue (£)')
axes[0].set_ylabel('')

sns.barplot(data=top_quantity, x='QuantitySold', y='Product',
            palette='Blues_d', ax=axes[1])
axes[1].set_title('Top 10 Products by Quantity Sold', fontsize=13)
axes[1].set_xlabel('Total Quantity Sold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()

print("=== Top 10 by Revenue ===")
print(top_revenue)
print("\n=== Top 10 by Quantity ===")
print(top_quantity)

# Business overview summary
total_revenue       = df['TotalPrice'].sum()
total_orders        = df['InvoiceNo'].nunique()
total_customers     = df['CustomerID'].nunique()
total_products      = df['Description'].nunique()
avg_order_value     = df.groupby('InvoiceNo')['TotalPrice'].sum().mean()
avg_items_per_order = df.groupby('InvoiceNo')['Quantity'].sum().mean()

print("=" * 40)
print("       BUSINESS OVERVIEW")
print("=" * 40)
print(f"Total Revenue:         £{total_revenue:,.2f}")
print(f"Total Orders:          {total_orders:,}")
print(f"Total Customers:       {total_customers:,}")
print(f"Total Products:        {total_products:,}")
print(f"Avg Order Value:       £{avg_order_value:,.2f}")
print(f"Avg Items per Order:   {avg_items_per_order:,.1f}")
print(f"Date Range:            {df['InvoiceDate'].min().date()} → {df['InvoiceDate'].max().date()}")
print("=" * 40)

# ── 5. Customer-Level Analysis ────────────────────────────────────────────
# I collapse all transactions to one row per customer to analyze behavior

customer_df = df.groupby('CustomerID').agg(
    TotalRevenue  = ('TotalPrice',  'sum'),
    TotalOrders   = ('InvoiceNo',   'nunique'),
    TotalItems    = ('Quantity',    'sum'),
    AvgOrderValue = ('TotalPrice',  'mean'),
    FirstPurchase = ('InvoiceDate', 'min'),
    LastPurchase  = ('InvoiceDate', 'max'),
).reset_index()

# Days between first and last purchase — a proxy for customer loyalty lifespan
customer_df['CustomerLifespanDays'] = (
    customer_df['LastPurchase'] - customer_df['FirstPurchase']
).dt.days

print(f"Customer dataframe shape: {customer_df.shape}")
print("\nSample:")
print(customer_df.head())
print("\nStats:")
print(customer_df[['TotalRevenue','TotalOrders','TotalItems','CustomerLifespanDays']].describe())

# Pareto analysis — verifying whether ~20% of customers drive ~80% of revenue
customer_sorted = customer_df.sort_values('TotalRevenue', ascending=False).reset_index(drop=True)

customer_sorted['CumulativeRevenue']    = customer_sorted['TotalRevenue'].cumsum()
customer_sorted['CumulativeRevenuePct'] = (customer_sorted['CumulativeRevenue']
                                            / customer_sorted['TotalRevenue'].sum() * 100)
customer_sorted['CustomerPct'] = (customer_sorted.index + 1) / len(customer_sorted) * 100

cutoff_80  = customer_sorted[customer_sorted['CumulativeRevenuePct'] >= 80].iloc[0]
cutoff_idx = cutoff_80.name
pct_customers = (cutoff_idx + 1) / len(customer_sorted) * 100

print(f"Top {pct_customers:.1f}% of customers generate 80% of revenue")
print(f"That's {cutoff_idx + 1} customers out of {len(customer_sorted)}")

plt.figure(figsize=(10, 5))
plt.plot(customer_sorted['CustomerPct'],
         customer_sorted['CumulativeRevenuePct'],
         color='steelblue', linewidth=2)
plt.axhline(y=80, color='red',    linestyle='--', alpha=0.7, label='80% revenue')
plt.axvline(x=pct_customers, color='orange', linestyle='--', alpha=0.7,
            label=f'Top {pct_customers:.1f}% of customers')
plt.title('Pareto Curve — Cumulative Revenue by Customer', fontsize=13)
plt.xlabel('Cumulative % of Customers')
plt.ylabel('Cumulative % of Revenue')
plt.legend()
plt.tight_layout()
plt.show()

# Revenue distribution — I show both the full distribution and a trimmed version
# (top 1% removed) so the shape of the main customer base is visible
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].hist(customer_df['TotalRevenue'], bins=100, color='steelblue', edgecolor='none')
axes[0].set_title('Customer Revenue Distribution (Full)', fontsize=13)
axes[0].set_xlabel('Total Revenue (£)')
axes[0].set_ylabel('Number of Customers')

p99     = customer_df['TotalRevenue'].quantile(0.99)
trimmed = customer_df[customer_df['TotalRevenue'] <= p99]

axes[1].hist(trimmed['TotalRevenue'], bins=100, color='steelblue', edgecolor='none')
axes[1].set_title('Customer Revenue Distribution (Top 1% Removed)', fontsize=13)
axes[1].set_xlabel('Total Revenue (£)')
axes[1].set_ylabel('Number of Customers')

plt.tight_layout()
plt.show()

print("=== Revenue Brackets ===")
brackets = [0, 100, 500, 1000, 5000, 10000, float('inf')]
labels   = ['£0-100', '£100-500', '£500-1k', '£1k-5k', '£5k-10k', '£10k+']

customer_df['RevenueBracket'] = pd.cut(customer_df['TotalRevenue'],
                                        bins=brackets, labels=labels)
bracket_counts = customer_df['RevenueBracket'].value_counts().sort_index()
bracket_pct    = (bracket_counts / len(customer_df) * 100).round(1)

for label, count, pct in zip(labels, bracket_counts, bracket_pct):
    print(f"  {label:<12}  {count:>4} customers  ({pct}%)")

# Purchase frequency — I cap at 20 orders for chart readability,
# and compare average revenue across frequency brackets to show
# how repeat behavior drives value
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

order_freq = customer_df['TotalOrders'].clip(upper=20)
axes[0].hist(order_freq, bins=20, color='steelblue', edgecolor='white')
axes[0].set_title('Distribution of Orders per Customer', fontsize=13)
axes[0].set_xlabel('Number of Orders (capped at 20)')
axes[0].set_ylabel('Number of Customers')

one_time = (customer_df['TotalOrders'] == 1).sum()
repeat   = (customer_df['TotalOrders'] >  1).sum()

axes[1].pie([one_time, repeat],
            labels=[f'One-time buyers\n({one_time:,})',
                    f'Repeat buyers\n({repeat:,})'],
            colors=['#5b9bd5', '#2e75b6'],
            autopct='%1.1f%%',
            startangle=90,
            textprops={'fontsize': 12})
axes[1].set_title('One-time vs Repeat Buyers', fontsize=13)

plt.tight_layout()
plt.show()

print("=== Purchase Frequency Breakdown ===")
freq_brackets = [1, 2, 3, 5, 10, 20, float('inf')]
freq_labels   = ['1 order', '2 orders', '3-4 orders',
                 '5-9 orders', '10-19 orders', '20+ orders']

customer_df['FreqBracket'] = pd.cut(customer_df['TotalOrders'],
                                     bins=freq_brackets,
                                     labels=freq_labels,
                                     right=False)
freq_counts = customer_df['FreqBracket'].value_counts().sort_index()

for label, count in zip(freq_labels, freq_counts):
    pct     = count / len(customer_df) * 100
    avg_rev = customer_df[customer_df['FreqBracket']==label]['TotalRevenue'].mean()
    print(f"  {label:<15} {count:>4} customers ({pct:4.1f}%)  "
          f"avg revenue: £{avg_rev:,.0f}")

# ── 6. RFM Analysis ───────────────────────────────────────────────────────
# RFM scores each customer on three dimensions:
#   Recency   — days since last purchase (lower = better)
#   Frequency — number of orders placed (higher = better)
#   Monetary  — total spend (higher = better)

import datetime

# I use the day after the last transaction as the reference "today"
reference_date = df['InvoiceDate'].max() + datetime.timedelta(days=1)
print(f"Reference date: {reference_date.date()}")

rfm = df.groupby('CustomerID').agg(
    Recency   = ('InvoiceDate', lambda x: (reference_date - x.max()).days),
    Frequency = ('InvoiceNo',   'nunique'),
    Monetary  = ('TotalPrice',  'sum')
).reset_index()

print(f"\nRFM dataframe shape: {rfm.shape}")
print("\nSample:")
print(rfm.head(10))
print("\nRFM Stats:")
print(rfm[['Recency','Frequency','Monetary']].describe())

# Score each dimension 1-5 using quintiles
# Recency is reversed — a lower recency (bought recently) gets a higher score
rfm['R_Score'] = pd.qcut(rfm['Recency'],
                          q=5,
                          labels=[5,4,3,2,1],
                          duplicates='drop')

# rank(method='first') breaks ties in Frequency before binning
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'),
                          q=5,
                          labels=[1,2,3,4,5],
                          duplicates='drop')

rfm['M_Score'] = pd.qcut(rfm['Monetary'],
                          q=5,
                          labels=[1,2,3,4,5],
                          duplicates='drop')

rfm['R_Score'] = rfm['R_Score'].astype(int)
rfm['F_Score'] = rfm['F_Score'].astype(int)
rfm['M_Score'] = rfm['M_Score'].astype(int)

rfm['RFM_Score']   = rfm['R_Score'] + rfm['F_Score'] + rfm['M_Score']
rfm['RFM_Segment'] = (rfm['R_Score'].astype(str) +
                       rfm['F_Score'].astype(str) +
                       rfm['M_Score'].astype(str))

print("RFM Score distribution:")
print(rfm['RFM_Score'].describe())
print("\nSample with scores:")
print(rfm[['CustomerID','Recency','Frequency','Monetary',
           'R_Score','F_Score','M_Score','RFM_Score']].head(10))

# Assign business segment labels based on R and F score combinations
def assign_segment(row):
    r = row['R_Score']
    f = row['F_Score']
    m = row['M_Score']

    if r >= 4 and f >= 4:               return 'Champions'
    elif r >= 3 and f >= 3:             return 'Loyal Customers'
    elif r >= 4 and f <= 2:             return 'New Customers'
    elif r >= 3 and f <= 2:             return 'Promising'
    elif r <= 2 and f >= 3:             return 'At Risk'
    elif r <= 2 and f >= 4:             return 'Can\'t Lose Them'
    elif r == 1 and f == 1 and m >= 4:  return 'Big Spender Lost'
    elif r <= 2 and f <= 2:             return 'Lost'
    else:                               return 'Needs Attention'

rfm['Segment'] = rfm.apply(assign_segment, axis=1)

segment_summary = rfm.groupby('Segment').agg(
    CustomerCount = ('CustomerID',  'count'),
    AvgRecency    = ('Recency',     'mean'),
    AvgFrequency  = ('Frequency',   'mean'),
    AvgMonetary   = ('Monetary',    'mean'),
    TotalRevenue  = ('Monetary',    'sum')
).round(1).reset_index()

segment_summary['RevenuePct'] = (segment_summary['TotalRevenue'] /
                                  segment_summary['TotalRevenue'].sum() * 100).round(1)

segment_summary = segment_summary.sort_values('TotalRevenue', ascending=False)
print(segment_summary.to_string(index=False))

# Segment visualizations — customer count, revenue, and R vs F scatter
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

colors = {
    'Champions':       '#1a5276',
    'Loyal Customers': '#2e86c1',
    'At Risk':         '#e67e22',
    'Lost':            '#c0392b',
    'Promising':       '#27ae60',
    'New Customers':   '#2ecc71',
    'Big Spender Lost':'#8e44ad',
    'Needs Attention': '#95a5a6'
}

segment_summary_sorted = segment_summary.sort_values('CustomerCount', ascending=False)
bar_colors = [colors.get(s, '#95a5a6') for s in segment_summary_sorted['Segment']]

axes[0].barh(segment_summary_sorted['Segment'],
             segment_summary_sorted['CustomerCount'],
             color=bar_colors)
axes[0].set_title('Customers per Segment', fontsize=13)
axes[0].set_xlabel('Number of Customers')
axes[0].invert_yaxis()

segment_rev = segment_summary.sort_values('TotalRevenue', ascending=False)
rev_colors  = [colors.get(s, '#95a5a6') for s in segment_rev['Segment']]
axes[1].barh(segment_rev['Segment'],
             segment_rev['TotalRevenue'],
             color=rev_colors)
axes[1].set_title('Total Revenue per Segment', fontsize=13)
axes[1].set_xlabel('Total Revenue (£)')
axes[1].invert_yaxis()

for segment, group in rfm.groupby('Segment'):
    color = colors.get(segment, '#95a5a6')
    axes[2].scatter(group['Recency'],
                    group['Frequency'].clip(upper=30),
                    c=color, label=segment, alpha=0.5, s=20)

axes[2].set_title('Recency vs Frequency by Segment', fontsize=13)
axes[2].set_xlabel('Recency (days since last purchase)')
axes[2].set_ylabel('Frequency (orders, capped at 30)')
axes[2].legend(fontsize=8, loc='upper right')

plt.tight_layout()
plt.show()

# ── 7. Save Processed Data ────────────────────────────────────────────────

save_path = '/content/drive/MyDrive/data/'

df.to_csv(save_path          + 'online_retail_clean.csv', index=False)
rfm.to_csv(save_path         + 'rfm_segments.csv',        index=False)
customer_df.to_csv(save_path + 'customer_profiles.csv',   index=False)

print("Saved:")
print(f"  online_retail_clean.csv  — {len(df):,} rows")
print(f"  rfm_segments.csv         — {len(rfm):,} rows")
print(f"  customer_profiles.csv    — {len(customer_df):,} rows")

# ── 8. Cohort Analysis ────────────────────────────────────────────────────
# I group customers by their first purchase month (cohort) and track
# what percentage of each cohort returned in subsequent months

df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')

# Find first purchase month per customer, then merge back onto transactions
cohort_map = (df.groupby('CustomerID')['InvoiceMonth']
                .min()
                .reset_index())
cohort_map.columns = ['CustomerID', 'CohortMonth']

df = df.merge(cohort_map, on='CustomerID', how='left')

# CohortIndex = how many months after first purchase this transaction occurred
df['CohortIndex'] = (df['InvoiceMonth'] - df['CohortMonth']).apply(lambda x: x.n)

print("Cohort index sample:")
print(df[['CustomerID','InvoiceMonth','CohortMonth','CohortIndex']].head(10))
print(f"\nMax cohort index: {df['CohortIndex'].max()} months")

# Count unique active customers per cohort/month combination
cohort_data = (df.groupby(['CohortMonth', 'CohortIndex'])['CustomerID']
                 .nunique()
                 .reset_index())
cohort_data.columns = ['CohortMonth', 'CohortIndex', 'Customers']

# Pivot to a matrix, then divide each row by its cohort's starting size
cohort_pivot = cohort_data.pivot_table(index='CohortMonth',
                                        columns='CohortIndex',
                                        values='Customers')
cohort_sizes = cohort_pivot[0]
retention    = cohort_pivot.divide(cohort_sizes, axis=0).round(3) * 100

print("Cohort sizes (new customers per month):")
print(cohort_sizes)
print("\nRetention table (% returning each month):")
print(retention.round(1))

# Retention heatmap — color intensity shows retention rate per cell
import numpy as np

fig, ax = plt.subplots(figsize=(16, 8))
mask = retention.isnull()

sns.heatmap(retention,
            annot=True,
            fmt='.0f',
            mask=mask,
            cmap='Blues',
            vmin=0,
            vmax=60,
            linewidths=0.5,
            linecolor='white',
            ax=ax,
            cbar_kws={'label': 'Retention Rate (%)'})

ax.set_title('Monthly Cohort Retention Rate (%)\n'
             'Each cell = % of cohort still purchasing that month',
             fontsize=14, pad=15)
ax.set_xlabel('Months Since First Purchase', fontsize=12)
ax.set_ylabel('Cohort (First Purchase Month)', fontsize=12)
ax.set_yticklabels([str(m) for m in retention.index], rotation=0)

plt.tight_layout()
plt.show()

print("=== Key Retention Metrics ===")
print(f"\nMonth 1 retention by cohort:")
for month, rate in retention[1].dropna().items():
    print(f"  {month}:  {rate:.1f}%")

avg_m1 = retention[1].dropna().mean()
print(f"\nAverage Month 1 retention: {avg_m1:.1f}%")
print(f"Average Month 3 retention: {retention[3].dropna().mean():.1f}%")
print(f"Average Month 6 retention: {retention[6].dropna().mean():.1f}%")

# ── 9. Final Summary ──────────────────────────────────────────────────────

print("=" * 55)
print("   E-COMMERCE CUSTOMER BEHAVIOR ANALYSIS")
print("   Final Summary of Key Findings")
print("=" * 55)

print("""
DATASET
  Source:      UCI Online Retail Dataset
  Period:      Dec 2010 — Dec 2011
  Raw rows:    541,909
  Clean rows:  397,884  (26.6% removed)
  Customers:   4,338
  Revenue:     £8,911,408
""")

print("─" * 55)
print("BUSINESS TYPE")
print("""
  B2B wholesale gift supplier based in UK.
  Evidence: avg order value £481, avg 279 items/order,
  zero Saturday orders, peak activity 10am-2pm weekdays.
""")

print("─" * 55)
print("REVENUE CONCENTRATION")
print(f"""
  Top 26% of customers → 80% of revenue (Pareto)
  Champions segment (26%): avg spend £5,204, avg 10 orders
  Lost segment (24%):      avg spend £395,  avg 1.1 orders

  Revenue by country: UK 82%, Netherlands 3.2%,
  Ireland 3.0%, Germany 2.6%, France 2.3%
""")

print("─" * 55)
print("SEASONALITY")
print("""
  Strong Christmas peak: revenue nearly doubles
  Jun-Aug (£600k/mo) → Nov 2011 (£1.16M)
  Post-holiday dip in Jan-Feb as expected
  December 2011 appears low — incomplete month only
""")

print("─" * 55)
print("CUSTOMER SEGMENTS  (RFM Analysis)")

for _, row in segment_summary.iterrows():
    print(f"  {row['Segment']:<20} "
          f"{row['CustomerCount']:>4} customers  "
          f"{row['RevenuePct']:>5.1f}% revenue  "
          f"avg £{row['AvgMonetary']:>7,.0f}")

print()
print("─" * 55)
print("RETENTION  (Cohort Analysis)")
print(f"""
  Average Month 1 retention:  20.6%
  Average Month 3 retention:  23.2%
  Average Month 6 retention:  24.4%

  Pattern: leaky bucket — ~79% of new customers
  never return after first purchase. The 21% who
  do return become stable long-term buyers.

  Dec 2010 cohort strongest: 36.6% month-1 retention,
  stable 33-40% throughout year, 50% spike in Nov 2011
  (Christmas reactivation effect).
""")

print("─" * 55)
print("TOP BUSINESS RECOMMENDATIONS")
print("""
  1. Protect Champions — 1,139 customers drive 66.5%
     of revenue. Any churn here is catastrophic.

  2. Win back At-Risk — 643 customers who bought
     regularly but went quiet ~5 months ago. They
     know the product. A targeted offer could recover 9%
     of revenue that is actively slipping away.

  3. Convert first-timers — getting a one-time buyer
     to order twice doubles their expected lifetime value
     from £413 to £886. This is the highest-leverage
     single action available.

  4. Investigate Big Spender Lost — 21 customers who
     each spent ~£5,000 and vanished. Personal outreach
     justified given the value at stake.
""")
print("=" * 55)
```
